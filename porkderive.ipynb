```python
import pandas as pd 
import re
```

```python
df = pd.read_csv("/ML/datasets/cleaned_dataset.csv")
```

```python
df = df[['text', 'label']]
```

```python
sneaky_keywords = [
    r'\bpork\b', r'\blard\b', r'\bbacon\b', r'\bham\b',      
    r'\bcarmine\b', r'\bcochineal\b', r'natural red 4',      
    r'\banimal fat\b', r'\brennet\b', r'\bpepsin\b',         
    r'\bgelatin\b', r'e120'                                 
]

pattern = re.compile('|'.join(sneaky_keywords), flags=re.IGNORECASE)
```

```python
sneaky_pork = df[df['text'].str.contains(pattern, na=False)].copy()

sneaky_pork['sneaky_found'] = sneaky_pork['text'].apply(
    lambda x: ', '.join(set(re.findall(pattern, str(x).lower())))
)
```

```python
print(f"Total sneaky items caught and verified: {len(sneaky_pork)}")

pd.set_option('display.max_colwidth', None)
display(sneaky_pork[['text', 'label', 'sneaky_found']].head(10))
```
